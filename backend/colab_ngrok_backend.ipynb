```python
image_folder = "/content/drive/MyDrive/AI COHORT/images/necklace" #path of the image folder inside the drive

image_paths = [os.path.join(image_folder, f) for f in os.listdir(image_folder) if f.endswith(".jpg")] #add image paths into list
```

```python
# Generate embeddings
image_embeddings = fclip.encode_images(image_paths, batch_size=32)

print("Total images:", len(image_embeddings))
print("Embedding size:", image_embeddings.shape)
```

```python
import chromadb

client = chromadb.Client()

collection = client.get_or_create_collection(
    name="jewelry_collection"
)

ids = []
metadatas = []

for path in local_path:
    image_name = path.replace("/necklace/","")
    image_id = image_name.replace(".jpg", "")

    ids.append(image_id)

    metadatas.append({
        "image_name": image_name,
        "category": image_name.replace(".jpg","").split('_')[0],
        "image_path": path
    })

collection.add(
    ids=ids,
    embeddings=image_embeddings.tolist(),
    metadatas=metadatas
)

print("✅ Added to vector DB")
```

#HANDWRITTEN --> IMAGE

```python
from transformers import TrOCRProcessor, VisionEncoderDecoderModel
from PIL import Image  # You need to import PIL.Image
import torch

# Set device
device = "cuda" if torch.cuda.is_available() else "cpu"

# Load processor and model
processor = TrOCRProcessor.from_pretrained("microsoft/trocr-large-handwritten")
model = VisionEncoderDecoderModel.from_pretrained("microsoft/trocr-large-handwritten")

# Move model to device
model = model.to(device)

```

```python
from PIL import Image, ImageEnhance
import numpy as np
import cv2
# Load image
image_path = "/content/drive/MyDrive/AI COHORT/images/WhatsApp Image 2026-02-08 at 10.26.51 PM.jpeg"
image = Image.open(image_path).convert("RGB")
# Preprocessing pipeline
img_array = np.array(image)
gray = cv2.cvtColor(img_array, cv2.COLOR_RGB2GRAY)

# Apply Gaussian blur to reduce noise
blurred = cv2.GaussianBlur(gray, (5, 5), 0)

# Adaptive thresholding
thresh = cv2.adaptiveThreshold(blurred, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
                                cv2.THRESH_BINARY, 11, 2)

# Convert back to RGB
image_processed = Image.fromarray(thresh).convert("RGB")

# Enhance contrast
enhancer = ImageEnhance.Contrast(image_processed)
image_processed = enhancer.enhance(2.5)

# Process image and move to device
pixel_values = processor(images=image_processed, return_tensors="pt").pixel_values.to(device)

# Generate text
generated_ids = model.generate(pixel_values)
generated_text = processor.batch_decode(generated_ids, skip_special_tokens=True)[0]

print(f"Extracted Text: {generated_text}")
```

```python
from flask import Flask, request, jsonify, send_file
from flask_cors import CORS
import os
from PIL import Image, ImageEnhance
import numpy as np
import cv2
```

```python
app = Flask(__name__)
CORS(app)

BASE_IMAGE_PATH = "/content/drive/MyDrive/AI COHORT/images"
```

```python
@app.route("/images/necklace/<filename>")
def serve_image(category, filename):
    file_path = os.path.join(BASE_IMAGE_PATH, category, filename)
    return send_file(file_path, mimetype="image/jpeg")
```

```python
def format_results(results_meta):
    base_url = request.host_url
    formatted = []

    for item in results_meta:
        formatted.append({
            "category": item["category"],
            "image_name": item["image_name"],
            "image_url": item['image_path']
        })

    return formatted
```

```python
@app.route("/textSearch", methods=["POST"])
def text_search():
    data = request.json
    query = data.get("query")

    text_embedding = fclip.encode_text([query], batch_size=1)

    results = collection.query(
        query_embeddings=text_embedding.tolist(),
        n_results=10
    )

    formatted = format_results(results["metadatas"][0])
    print(results)

    return jsonify({"results": formatted})
```

```python
@app.route("/imageSearch", methods=["POST"])
def image_search():
    if 'file' not in request.files:
        return jsonify({"error": "No file uploaded"}), 400

    file = request.files['file']
    temp_path = "temp.jpg"
    file.save(temp_path)

    query_embedding = fclip.encode_images([temp_path], batch_size=1)

    results = collection.query(
        query_embeddings=query_embedding.tolist(),
        n_results=10
    )

    os.remove(temp_path)

    formatted = format_results(results["metadatas"][0])

    return jsonify({"results": formatted})
```

```python
@app.route("/handwritingSearch", methods=["POST"])
def handwriting_search():

    if 'file' not in request.files:
        return jsonify({"error": "No file uploaded"}), 400

    file = request.files['file']
    temp_path = "hand.jpg"
    file.save(temp_path)

    image = Image.open(temp_path).convert("RGB")

    img_array = np.array(image)
    gray = cv2.cvtColor(img_array, cv2.COLOR_RGB2GRAY)
    blurred = cv2.GaussianBlur(gray, (5, 5), 0)

    thresh = cv2.adaptiveThreshold(
        blurred, 255,
        cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
        cv2.THRESH_BINARY, 11, 2
    )

    image_processed = Image.fromarray(thresh).convert("RGB")

    enhancer = ImageEnhance.Contrast(image_processed)
    image_processed = enhancer.enhance(2.5)

    pixel_values = processor(images=image_processed,
                             return_tensors="pt").pixel_values.to(device)

    generated_ids = model.generate(pixel_values)
    generated_text = processor.batch_decode(
        generated_ids, skip_special_tokens=True)[0]

    text_embedding = fclip.encode_text([generated_text], batch_size=1)

    results = collection.query(
        query_embeddings=text_embedding.tolist(),
        n_results=10
    )

    os.remove(temp_path)

    formatted = format_results(results["metadatas"][0])

    return jsonify({
        "extracted_text": generated_text,
        "results": formatted
    })
```

```python
@app.route("/drawingSearch", methods=["POST"])
def drawing_search():
    if 'file' not in request.files:
        return jsonify({"error": "No file uploaded"}), 400

    file = request.files['file']
    temp_path = "drawing.jpg"
    file.save(temp_path)

    query_embedding = fclip.encode_images([temp_path], batch_size=1)

    results = collection.query(
        query_embeddings=query_embedding.tolist(),
        n_results=10
    )

    os.remove(temp_path)

    formatted = format_results(results["metadatas"][0])

    return jsonify({"results": formatted})
```
